##Plaintext vs Cyphertext
* Plaintext means a chunk of unencrypted text meant to be eventually ecrypted use a cryptography algorithm or tool. These messages can be read by anyone with access to the network or device. Eavesdroppers can read plaintext directly without the need for any cryptographic tools.
* Cyphertext means a chunk of text that has been run through some kind of cypher to cryptographically hide the content from those without the keys or tools to decrypt the original message. The message is garbled nonsense to someone eavesdropping on a network or device who doesn't have the correct cryptographic key or tool.

##One-Time Pad Background
The one-time pad (OTP) is a symmetric key encryption method known for its perfect secrecy. It requires a key that is at least as long as the message being encrypted. This technique ensures that each bit of the plaintext is securely combined with a corresponding bit from the key.


**Key Characteristics**
* Key Length: The key must be equal to or longer than the plaintext message.
* Randomness: The key must be truly random to ensure security.
* Single Use: Each key can only be used once; reusing keys compromises security.
* Secrecy: The key must remain completely secret between the communicating parties.

**How the One-Time Page Works**

The encryption process involves the following steps:

1. Key Generation: A random key is generated, matching the length of the plaintext.
2. Encryption: Each bit of the plaintext is combined with the corresponding bit of the key using modular addition (often represented as XOR in binary).
3. Ciphertext Creation: The result of the combination is the ciphertext, which appears random and bears no resemblance to the original message.


##Example of Encryption

| Plaintext Character | Key Character | Resulting Ciphertext Character |
|:-------------------:|:-------------:|:------------------------------:|
| **A**                   |	**R**             |	**(A + R) mod 256**                |
| **B**                   | **T**             |	**(B + T) mod 256**                |
| **C**                   | **Z**             |	**(C + Z) mod 256**                |


Security and Limitations

The one-time pad is theoretically unbreakable under the following conditions:
* The key is truly random.
* The key is used only once.
* The key is kept secret.

*However, practical challenges arise in key distribution and management, making the one-time pad less feasible for everyday use compared to modern encryption methods.*

#Alice generates a secret key.

Define a string called secretKeyAlice of length n that will be composed of a random sequence of '0' and '1' characters. We are using a size of 8-bits based on the ASCII standard. In this example, each 8-bit character will be represented by a qubit.

```python
#@title Setup Alice Message Variables
# Random is used to choose random values or generate random numbers
import random

# The initial number of characters in our text message we want to send.
MESSAGE_LENGTH = 10

# The number of qubits we will use.
# We multiply by 8 because each character is represented by 8 bits in ASCII.
n = 8 * MESSAGE_LENGTH

# This empty string will become ALice's secret key through iteration.
secretKeyAlice = ""
```

## Convert the keys from ASCII into Binary
Here we convert the keys for bob and alice from ASCII encoded text into actual binary with 8-bits per character. If you're curious what this looks like remove the comment character (`#') from in front of the print statements and run the block again.



```python
secretKeyBob_bin = ''.join(format(ord(char), '08b') for char in secretKeyBob)
secretKeyAlice_bin = ''.join(format(ord(char), '08b') for char in secretKeyAlice)
print("secretKeyBob_bin \t= " + secretKeyBob_bin)
#print("secretKeyAlice_bin \t= "+ secretKeyAlice_bin)
```

```python
numMatch = 0

if len(secretKeyAlice_bin) != len(secretKeyBob_bin):
    print("Secret keys are different lengths!")
else:
    for i in range(len(secretKeyAlice_bin)):
        if secretKeyAlice_bin[i] == secretKeyBob_bin[i]:
           numMatch += 1
    matchPercent = numMatch / len(secretKeyAlice_bin) * 100

print(str(matchPercent) + "% match")

# Sample Output:
# 100.0% match
```
